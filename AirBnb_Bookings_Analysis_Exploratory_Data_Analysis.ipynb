### Are private rooms preferred over other room ?

````python
# importing Library
import pandas as pd


# Grouping by room type, then calculating total reviews and average price for each room type
room_type_stats = New_Airbnb_df.groupby('room_type').agg(
    total_reviews=('number_of_reviews', 'sum'),
    average_price=('price', 'mean')
).reset_index()

# Rounding the average_price to two decimal places for better readability
room_type_stats['average_price'] = room_type_stats['average_price'].round(2)

# Print the columns to check for errors
print("Room Type Stats Columns:", room_type_stats.columns)

# Extracting the statistics for private rooms
private_room_stats = room_type_stats.loc[room_type_stats['room_type'] == 'Private room'].iloc[0]

# Initializing a list to collect comparison results
comparisons = []

# Comparing private room with other room types
for _, row in room_type_stats.iterrows():
    if row['room_type'] != 'Private room':
        comparison_result = {
            'room_type': row['room_type'],
            'private_room_reviews': private_room_stats['total_reviews'],
            'other_room_reviews': row['total_reviews'],
            'price_difference': private_room_stats['average_price'] - row['average_price'],
            'reviews_difference': private_room_stats['total_reviews'] - row['total_reviews'],
            'private_room_price': private_room_stats['average_price'],  # Add private room price for clarity
            'other_room_price': row['average_price']  # Add other room price for clarity
        }
        comparisons.append(comparison_result)

# Converting comparisons to DataFrame for easier analysis
comparisons_df = pd.DataFrame(comparisons)

# Displaying the comparison results
print("Comparison of Private Room with Other Room Types:")
for index, row in comparisons_df.iterrows():
    print(f"Private Room vs. {row['room_type']}:")
    print(f"  Total Reviews - Private Room: {row['private_room_reviews']}, Other: {row['other_room_reviews']}")
    print(f"  Average Price - Private Room: ${row['private_room_price']:.2f}, Other: ${row['other_room_price']:.2f}")
    print(f"  Price Difference: ${row['price_difference']:.2f}")
    print(f"  Reviews Difference: {row['reviews_difference']}\n")
````

````python

# Calculate the average minimum nights within each room type group
average_minimum_nights = rooms_df.groupby('room_type')['minimum_nights'].mean().round(2).sort_values(ascending=False)

# Create a pivot table with average price
pivot_table_rooms_avg_price = pd.pivot_table(rooms_df, values='price', index='room_type', aggfunc=lambda x: round(x.mean(), 2), fill_value=0)

# Joining the two tables
combined_table = pivot_table_rooms_avg_price.join(average_minimum_nights, on='room_type', rsuffix='_avg_min_nights')

# Rename columns
combined_table.rename(
    columns={'price': 'Average Price', 'minimum_nights': 'Average Minimum Nights'},
    inplace=True
)
print("Combined Table:")
print(combined_table)
````

````python
# Analysis code
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Function to remove outliers
def remove_outliers(df, column, group_col):
    # Calculate the IQR for each group
    Q1 = df.groupby(group_col)[column].quantile(0.25)
    Q3 = df.groupby(group_col)[column].quantile(0.75)
    IQR = Q3 - Q1

    # Define the acceptable range
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Filter the data to remove outliers
    def filter_outliers(row):
        if row[column] < lower_bound[row[group_col]] or row[column] > upper_bound[row[group_col]]:
            return False
        return True

    return df[df.apply(filter_outliers, axis=1)]

# Remove outliers in the 'price' column based on 'neighbourhood_group'
cleaned_data = remove_outliers(New_Airbnb_df, 'price', 'neighbourhood_group')
````

````python
# Chart - 7 visualization code for  code Monthly Growth Rate of Average Reviews (%)

import seaborn as sns
import matplotlib.pyplot as plt


# Convert the "last_review" column to datetime format
New_Airbnb_df['last_review'] = pd.to_datetime(New_Airbnb_df['last_review'])

# Extract the month from the "last_review" column
New_Airbnb_df['review_month'] = New_Airbnb_df['last_review'].dt.month

# Aggregate reviews by month and calculate the Average of reviews
Average_review_by_month = New_Airbnb_df.groupby('review_month')['number_of_reviews'].mean()

# Calculate month-over-month growth rate as a percentage
growth_rate_percentage = Average_review_by_month.pct_change() * 100

# Plot the growth rate as a percentage
plt.figure(figsize=(10, 6))

# Plotting with conditional color and bar labels
growth_rate_percentage.plot(kind='bar', color=growth_rate_percentage.apply(lambda x: 'green' if x >= 0 else 'red')) # lambda function
plt.title('Monthly Growth Rate of Average Reviews (%)')
plt.xlabel('Months of Year')
plt.ylabel('Growth Rate (%)')
plt.axhline(y=0, color='k', linestyle='--', linewidth=0.5)  # Adding horizontal line at y=0
plt.xticks(rotation=0)
plt.grid(axis='x')
plt.show()
````
